```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# Re-initialize llm to ensure it's a ChatGoogleGenerativeAI instance
llm = ChatGoogleGenerativeAI(
    #  model="google_genai:gemini-2.5-flash-lite"
    model="gemini-2.5-flash",
    temperature=0.2  # Lower values keep responses grounded and factual
)

# 6. Build the LCEL execution chain
# Extracts text content from retrieved elements to pass cleanly into the text template

# Re-define format_docs if it was overwritten or not in scope
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# Re-define rag_chain to use the correct llm instance
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | rag_prompt
    | llm
    | StrOutputParser()
)

# 7. Execute a test question through the complete pipeline
if query is None:
  user_question = "What is the main subject of this document? explain me any one poem which the writer tells"
else:
  user_question = query

print(f"User Question: {user_question}\n")

print("Processing query and generating answer...")
ai_response = rag_chain.invoke(user_question)

print("\n--- Gemini Answer ---")
print(ai_response)
```

```python
import gradio as gr

def process_rag_query(audio_path):
    # Transcribe the audio
    whisper_model = whisper.load_model("base") # Re-load model in the function for Gradio to handle gracefully
    transcription_result = whisper_model.transcribe(audio_path)
    user_query = transcription_result["text"]

    # Normalize query
    user_query = unicodedata.normalize("NFC", user_query)

    # Get answer from RAG chain
    ai_response = rag_chain.invoke(user_query)

    return user_query, ai_response


# Create the Gradio interface
iface = gr.Interface(
    fn=process_rag_query,
    inputs=gr.Audio(type="filepath", label="Upload your voice query"),
    outputs=[
        gr.Textbox(label="Transcribed Query"),
        gr.Textbox(label="RAG Answer")
    ],
    title="Kannada RAG with Voice Input",
    description="Upload an audio file with your question to get an answer from the RAG system."
)

# Launch the interface
iface.launch(debug=True)
```

## Updated Gradio Interface for Universal RAG

```python
import gradio as gr
import whisper
import unicodedata
import os
import shutil # For removing old Chroma DB

# Imports for PDF processing and RAG chain
import pymupdf4llm
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
from langdetect import detect
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

# Global variables to store models and vector_db state
whisper_model_global = None
embedding_model_global = None
vector_db_global = None
rag_chain_global = None

def init_models_and_db(pdf_file_path=None):
    global whisper_model_global, embedding_model_global, vector_db_global, rag_chain_global

    # Load Whisper model once
    if whisper_model_global is None:
        print("Initializing Whisper model...")
        whisper_model_global = whisper.load_model("base")
        print("Whisper model loaded.")

    # Load Embedding model once
    if embedding_model_global is None:
        print("Initializing Embedding model...")
        embedding_model_global = HuggingFaceEmbeddings(model_name="BAAI/bge-m3")
        print("Embedding model loaded.")

    # Process PDF and build/rebuild vector_db if a new PDF is provided
    if pdf_file_path:
        print(f"Processing new PDF: {pdf_file_path}")
        try:
            md_text = pymupdf4llm.to_markdown(pdf_file_path)
            # For this example, we'll process the text directly

            text_splitter = RecursiveCharacterTextSplitter(
                chunk_size=1000,
                chunk_overlap=100,
                separators=["\n\n", "\n", " ", ""]
            )
            chunks = text_splitter.split_text(md_text)

            documents = []
            for chunk in chunks:
                try:
                    lang = detect(chunk)
                except:
                    lang = "unknown"
                documents.append(Document(
                    page_content=unicodedata.normalize("NFC", chunk),
                    metadata={"language": lang}
                ))
            print(f"Created {len(documents)} document chunks from PDF.")

            persist_directory = "./chroma_db_gradio"
            if os.path.exists(persist_directory):
                shutil.rmtree(persist_directory) # Clear previous DB
                print(f"Removed old Chroma DB at {persist_directory}")

            vector_db_global = Chroma.from_documents(
                documents=documents,
                embedding=embedding_model_global,
                persist_directory=persist_directory
            )
            print("Vector database built successfully for the new PDF!")

            # Re-initialize retriever and rag_chain after building new DB
            retriever = vector_db_global.as_retriever(search_kwargs={"k": 5})

            rag_prompt = ChatPromptTemplate.from_messages([
                ("system",
                 """
You are a multilingual assistant.

Answer ONLY using the provided context.

Always answer in the SAME language as the user's question.

If the answer is not present in the context,
say you do not have enough information.

Context:
{context}
"""
                ),
                ("human", "{question}")
            ])

            llm = ChatGoogleGenerativeAI(
                model="gemini-2.5-flash",
                temperature=0.2
            )

            def format_docs(docs):
                return "\n\n".join(doc.page_content for doc in docs)

            rag_chain_global = (
                {"context": retriever | format_docs, "question": RunnablePassthrough()}
                | rag_prompt
                | llm
                | StrOutputParser()
            )
            print("RAG chain re-initialized with new vector database.")
            return "PDF processed and knowledge base updated! You can now ask questions."

        except Exception as e:
            print(f"Error processing PDF: {e}")
            return f"Error processing PDF: {e}"

    elif vector_db_global is None:
        # Try to load an existing DB if none was built and no new PDF uploaded
        persist_directory = "./chroma_db_gradio"
        if os.path.exists(persist_directory):
            print(f"Loading existing Chroma DB from {persist_directory}")
            vector_db_global = Chroma(persist_directory=persist_directory, embedding_function=embedding_model_global)
            print("Existing vector database loaded.")

            retriever = vector_db_global.as_retriever(search_kwargs={"k": 5})
            rag_prompt = ChatPromptTemplate.from_messages([
                ("system",
                 """
You are a multilingual assistant.

Answer ONLY using the provided context.

Always answer in the SAME language as the user's question.

If the answer is not present in the context,
say you do not have enough information.

Context:
{context}
"""
                ),
                ("human", "{question}")
            ])
            llm = ChatGoogleGenerativeAI(
                model="gemini-2.5-flash",
                temperature=0.2
            )
            def format_docs(docs):
                return "\n\n".join(doc.page_content for doc in docs)

            rag_chain_global = (
                {"context": retriever | format_docs, "question": RunnablePassthrough()}
                | rag_prompt
                | llm
                | StrOutputParser()
            )
            print("RAG chain initialized from existing vector database.")
            return "Existing knowledge base loaded. You can now ask questions."
        else:
            print("No existing vector database found.")
            return "Please upload a PDF document first to build the knowledge base."
    return "Knowledge base ready."


def process_rag_query_ui(pdf_file, audio_path, text_query_input):
    global whisper_model_global, embedding_model_global, vector_db_global, rag_chain_global

    # Ensure models and DB are initialized/updated
    status_message = init_models_and_db(pdf_file.name if pdf_file else None)

    if vector_db_global is None:
        return "", "", status_message # Query, Answer, Status

    user_query = ""
    if audio_path:
        print("Transcribing audio...")
        transcription_result = whisper_model_global.transcribe(audio_path)
        user_query = transcription_result["text"]
        print(f"Audio transcribed: {user_query}")
    elif text_query_input:
        user_query = text_query_input
        print(f"Using text query: {user_query}")

    if not user_query:
        return "", "", "Please provide a question either by voice or text." # Query, Answer, Status

    # Normalize query
    user_query = unicodedata.normalize("NFC", user_query)

    print("Invoking RAG chain...")
    try:
        ai_response = rag_chain_global.invoke(user_query)
        print("RAG chain invoked.")

        is_rag_answer = "do not have enough information" not in ai_response.lower()
        rag_indicator = "RAG Status: Answer based on provided document." if is_rag_answer else "RAG Status: No direct answer found in document."

        final_answer = ai_response + "\n\nPowered by Gemini 2.5"

        return user_query, final_answer, rag_indicator

    except Exception as e:
        print(f"Error during RAG chain invocation: {e}")
        return user_query, f"Error getting answer: {e}", "RAG Status: Error"


# Initialize models and DB once when the script starts (if an existing DB exists)
# This will try to load a pre-existing DB or just load the models.
# If a new PDF is uploaded later, init_models_and_db will handle rebuilding the DB.
init_models_and_db()


# Create the Gradio interface
iface = gr.Interface(
    fn=process_rag_query_ui,
    inputs=[
        gr.File(label="Upload PDF Document (Optional: Upload a new PDF to update knowledge base)"),
        gr.Audio(type="filepath", label="Speak your question here (optional)"),
        gr.Textbox(label="Or type your question here (optional)", placeholder="e.g., What is the main theme of this document?")
    ],
    outputs=[
        gr.Textbox(label="Your Query"),
        gr.Textbox(label="RAG Answer"),
        gr.Textbox(label="RAG Status")
    ],
    title="Universal RAG with Voice/Text Input",
    description="Upload a PDF document to build its knowledge base. Then, ask questions in any language using voice or text. Answers are powered by Gemini 2.5 and derived from the uploaded document."
)

# Launch the interface
iface.launch(debug=True, share=True)
```
