```python
pdfName = 'forRAG.pdf'
```

```python
md_text = pymupdf4llm.to_markdown(pdfName)
with open("output.md", "w", encoding="utf-8") as f:
    f.write(md_text)
```

```python
md_file_path = 'output.md'
with open(md_file_path, 'r', encoding='utf-8') as file:
    md_text = file.read()
```

```python
##used to split the text
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100,
    separators=["\n\n", "\n", " ", ""]
)

chunks = text_splitter.split_text(md_text)
```

```python
print(chunks[1])
```

```python
print(len(chunks[1]))
```

```python
# creating a vector db
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
print("embedding model = ",embedding_model)
```

```python
persist_directory = "./chroma_db_gemini"
```

```python
documents = []
for chunk in chunks:
  documents.append(Document(page_content=chunk))
```

```python
documents
```

```python
from langchain_community.vectorstores import Chroma
```

```python
print("Generating Gemini embeddings and building vector database...")
vector_db = Chroma.from_documents(
    documents=documents,
    embedding=embedding_model,
    persist_directory=persist_directory
)
print("Vector database built successfully!")
```

```python
vector_db
```

```python
# do the rag pipeline
```

```python
# 5. Test Query
query = "What is the main subject of this document?"
docs = vector_db.similarity_search(query, k=2)

print("\n--- Gemini Search Results ---")
for doc in docs:
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}\n")
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
# 5. Define an explicit Grounding System Prompt
# The prompt instructs Gemini to avoid hallucinating facts beyond the input chunks
rag_prompt = ChatPromptTemplate.from_messages([
    ("system", (
        "You are a helpful assistant answering questions based strictly on the provided context.\n"
        "If you do not know the answer or if it's not explicitly found in the context, state "
        "that you do not have enough information. Do not attempt to guess.\n\n"
        "--- Context ---\n{context}"
    )),
    ("human", "{question}")
])
```

```python
# 3. Configure the retriever instance
# 'k=3' tells Chroma to fetch the top 3 most relevant segments of text
retriever = vector_db.as_retriever(search_kwargs={"k": 3})
```

```python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# Re-initialize llm to ensure it's a ChatGoogleGenerativeAI instance
llm = ChatGoogleGenerativeAI(
    #  model="google_genai:gemini-2.5-flash-lite"
    model="gemini-2.5-flash",
    temperature=0.2  # Lower values keep responses grounded and factual
)

# 6. Build the LCEL execution chain
# Extracts text content from retrieved elements to pass cleanly into the text template

# Re-define format_docs if it was overwritten or not in scope
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# Re-define rag_chain to use the correct llm instance
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | rag_prompt
    | llm
    | StrOutputParser()
)

# 7. Execute a test question through the complete pipeline
user_question = "What is the main subject of this document?"
print(f"User Question: {user_question}\n")

print("Processing query and generating answer...")
ai_response = rag_chain.invoke(user_question)

print("\n--- Gemini Answer ---")
print(ai_response)
```
